# Final Project

## Fundamental Info

The exam consists on replicate the study in the following paper using alternative data. Part of the core work will be done in class and in groups.

All the group member must be present in class during the four days dedicated to the work on the project in class:
- 16th May
- 19th May
- 21st May
- 23rd May 

If motivated, only one absence is allowed.

If not all members of the group were present in class or the project was done at home, the professor will ask for an oral discussion, which can be done online.
