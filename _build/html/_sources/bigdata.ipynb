# Big Data Applications - The Course

This website contains materials and exercises for the Big Data Applications course in the Faculty of Economics, Politics and Social Sciences at the University of Bologna, taught by Professor Matteo Barigozzi.  
The slides provided were created by Professor Matteo Barigozzi, who is the course instructor.  
This site serves as a support tool for completing exercises in preparation for the exam of the course `96357 - Big Data Applications`, part of the Bachelor's degree program in Economics, Politics and Social Sciences (code 5819).

## Topics

The site includes examples covered in class as well as additional exercises to help you learn how to effectively use Python, R, and Matlab, and explore their capabilities.  
The sidebar lists the course topics organized by the lectures given by Professor Matteo Barigozzi.
